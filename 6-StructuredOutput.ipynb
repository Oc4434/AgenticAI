```python
from pydantic import BaseModel, Field

class Movie(BaseModel):

    title:str = Field(description = "The Title of the movie") 
    year:int  = Field(description= "The Year the MOvie Was Released")
    director:str = Field(description="The Director Of Movie")
    rating:float = Field(description= "The Rating Of movie out of 10")
    
```

```python
model_with_structure = model.with_structured_output(Movie)

model_with_structure
```

```python
response = model.invoke("Provide the detail about the movie Inception")

print(response.content)
```

```python
model_with_structure = model.with_structured_output(
    Movie,
    method="json_schema"
)

response = model_with_structure.invoke(
    "Provide a description of the Ramayana epic."
)

print(response)
```
