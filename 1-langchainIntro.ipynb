Generative AI Applications Using Agents AND OPENAI

LLM Model Integration with Google Gemini

```python
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
```

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("google_genai:gemini-3.5-flash-lite")

response = model.invoke("Write An Essay On Artificial Intelegence")

print(response.content)
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI 

model = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")

response = model.invoke("Write An Eassay for me on AI")
print(response.content)
```

GROQ Model Integration

```python
import os 
from langchain.chat_models import init_chat_model 

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY") 

model = init_chat_model("groq:openai/gpt-oss-120b")

response = model.invoke("What Is Mind Set Of SuperStar")

print(response.content)
```

```python
import os  
from langchain_groq import ChatGroq 

model = ChatGroq(model = "openai/gpt-oss-120b")

response = model.invoke("Why Do Parrot Talk")
print(response.content)
```

```python
model.invoke("Write me an 500 Words on AI")
```

```python
for chunk in model.stream("write 1000 words on money building business"):

    print(chunk.text , end = "|", flush=True)

```
