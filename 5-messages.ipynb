```python
###Detailed INfo To LLM with System Message 

from langchain.messages import SystemMessage, HumanMessage

system_msg = SystemMessage("""
Your Are Senior Python Developer with expertise in web Framework.
Always Provide code  Examples and explain your resoning Be Concise But Through in your Explanation
""")

message = [

    system_msg,
    HumanMessage("How Do I Create The Rest API?")
] 

response = model.invoke(message)

print(response.content)
```

```python
from langchain.messages import HumanMessage,AIMessage, SystemMessage

ai_msg = AIMessage("I'd be happy to help you with that Question")

message = [
    
    SystemMessage("You Are Helpful Assistant"),
    HumanMessage("Can You Help Me?"),
    ai_msg,
    HumanMessage("Great! what's 2+2")
]

response = model.invoke(message)

print(response.content)
```

```python
####===========Optional from cell 34 But Its Clean Method =========================####
from langchain_core.messages import (
    HumanMessage,
    AIMessage,
    ToolMessage
)

# 1. User message
human_msg = HumanMessage(
    content="What's the weather in San Francisco?"
)

# 2. AI requests a tool
ai_msg = AIMessage(
    content="",
    tool_calls=[
        {
            "name": "get_weather",
            "args": {"city": "San Francisco"},
            "id": "call_123",
            "type": "tool_call"
        }
    ]
)

# 3. Tool returns a result
tool_msg = ToolMessage(
    content="Temperature is 18°C and cloudy.",
    tool_call_id="call_123"
)

# 4. Send conversation history to LLM
message = [human_msg, ai_msg, tool_msg]

response = model.invoke(message)

print(response.content)
```
